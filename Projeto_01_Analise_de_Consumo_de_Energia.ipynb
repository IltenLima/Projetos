```python
#Dados do Consumo de Energia (kWh)
consumo = [120, 135, 110, 160, 175, 90, 80]

# 1 - Calcular consumo total
total = sum(consumo)

# 2 - Calcular consumo médio
media = total / len(consumo)

# 3 - Encontrar o maior consumo e o menor consumo
maior = max(consumo)
menor = min(consumo)

# 4 - Classificar o consumo médio
if media > 120:
    classificação = "Consumo elevado"
else:
    classificação = "Consumo dentro do limite"

# 5 - Exibir resultados
print("Consumo total:", total, "kWh")
print("Consumo médio:", media, "kWh")
print("Maior consumo:", maior, "kWh")
print("Menor consumo:", menor, "kWh")
print("Classificação:", classificação)
```

```python
# 6 - Criando limite
limite = 120

def analisar_consumo():
  total = sum(consumo)
  media = total / len(consumo)
  maior = max(consumo)
  menor = min(consumo)

  posicao = consumo.index(maior) + 1

  if media > limite:
    classificação = "Consumo elevado"
  else:
    classificação = "Consumo dentro do limite"

  print("Consumo total:", total, "kWh")
  print("Consumo médio:", media, "kWh")
  print("Maior consumo:", maior, "kWh")
  print("Menor consumo:", menor, "kWh")
  print("Posição do maior consumo:", posicao)
  print("Classificação:", classificação)

# 7 - Executando a função para analisar o consumo
analisar_consumo()
```
